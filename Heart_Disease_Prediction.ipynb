# Heart Disease Prediction — Mini Project

**Goal:** Build a simple educational machine-learning classification model using a small synthetic dataset.

> **Important:** This dataset is synthetic and tiny, so results are only for learning. This project is not a medical diagnostic tool.

## 1. Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

## 2. Load the dataset

```python
df = pd.read_csv('heart_disease_sample.csv')
df.head()
```

## 3. Understand and check the data

```python
print('Dataset shape:', df.shape)
print('\nMissing values per column:')
print(df.isnull().sum())
print('\nTarget counts:')
print(df['heart_disease'].value_counts())
df.describe()
```

## 4. Explore the data

```python
df['heart_disease'].value_counts().plot(kind='bar')
plt.title('Target Class Counts (Synthetic Sample)')
plt.xlabel('heart_disease (0 = no, 1 = yes)')
plt.ylabel('Number of rows')
plt.tight_layout()
plt.show()
```

## 5. Select features and target

```python
X = df[['age', 'sex', 'resting_bp', 'cholesterol']]
y = df['heart_disease']
print('Features:', list(X.columns))
print('Target: heart_disease')
```

## 6. Split data and train a model

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000)
)
model.fit(X_train, y_train)
print('Model training completed.')
```

## 7. Evaluate the model

```python
y_pred = model.predict(X_test)
print('Accuracy:', round(accuracy_score(y_test, y_pred), 2))
print('\nConfusion matrix:\n', confusion_matrix(y_test, y_pred))
print('\nClassification report:\n', classification_report(y_test, y_pred, zero_division=0))
```

## 8. Try one sample prediction

This is a demonstration only, not a health assessment.

```python
sample = pd.DataFrame([{
    'age': 50,
    'sex': 1,
    'resting_bp': 130,
    'cholesterol': 220
}])
prediction = model.predict(sample)[0]
print('Model output (0 = class 0, 1 = class 1):', int(prediction))
print('Do not use this output for medical decisions.')
```
